```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
```

```python
df = np.round(pd.read_csv('50_Startups.csv')[['R&D Spend', 'Administration', 'Marketing Spend', 'Profit']]/10000)
np.random.seed(9)

df = df.sample(5)
df
```

```python
df = df.iloc[:, 0:-1].copy()
df
```

```python
df.iloc[1, 0] = np.nan
df.iloc[3, 1] = np.nan
df.iloc[-1, -1] = np.nan
df
```

```python
df.head()
```

```python
# Step 1 - Impute all missing values with mean of respective col

df0 = df.copy()

for col in df.columns:
    df0[col] = df[col].fillna(df[col].mean())

df0
```

```python
# Remove the col1 imputed value
df1 = df0.copy()
df1.iloc[1,0] = np.nan
df1
```

```python
# Use 3 first rows to build a model and use the last for prediction
feature_columns = ['Administration', 'Marketing Spend']
X = df1.iloc[[0, 2, 3, 4], :][feature_columns]
X
```

```python
y = df1.iloc[[0, 2, 3, 4], :]['R&D Spend']
y
```

```python
lr = LinearRegression()
lr.fit(X, y)
feature_row = df1.iloc[[1]][feature_columns]
lr.predict(feature_row)
```
