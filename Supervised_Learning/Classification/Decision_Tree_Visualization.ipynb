### Prediction Path in Plain English

```python
# Function to explain the decision path taken by one sample
def explain_prediction_path(model, X, feature_names):

    # Find all the nodes that the sample passes through
    node_indicator = model.decision_path(X.reshape(1, -1))

    # Find the final leaf node where the sample ends
    leaf_id = model.apply(X.reshape(1, -1))

    # Get the IDs of all nodes visited by the sample
    node_ids = node_indicator.indices

    # Create an empty list to store the explanation
    explanation = []

    # Go through each node in the decision path
    for node_id in node_ids:

        # Stop when we reach the final leaf node
        if leaf_id[0] == node_id:
            break

        # Get the feature used for splitting at this node
        feature = model.tree_.feature[node_id]

        # Get the threshold used for the split
        threshold = model.tree_.threshold[node_id]

        # Get the actual feature name
        feature_name = feature_names[feature]

        # Get the actual value of this feature for our sample
        value = X[feature]

        # Check which side of the decision the sample goes
        if value <= threshold:

            # Sample goes to the left branch
            explanation.append(
                f"{feature_name} is {value:.2f}, "
                f"which is less than or equal to {threshold:.2f}"
            )

        else:

            # Sample goes to the right branch
            explanation.append(
                f"{feature_name} is {value:.2f}, "
                f"which is greater than {threshold:.2f}"
            )

    # Return all the decision steps as a single string
    return "\n".join(explanation)
```

```python
print(explain_prediction_path(clas, X, feature_names=iris.feature_names))
```

### Univariate Regression

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeRegressor
from dtreeviz import model

# Load the cars dataset
df_cars = pd.read_csv('cars.csv')

# Input feature
X = df_cars[['km_driven']]

# Target variable
y = df_cars['selling_price']

# Create Decision Tree Regressor
dt = DecisionTreeRegressor(
    max_depth=3,
    criterion='absolute_error',
    random_state=42
)

# Train the model
dt.fit(X, y)

# Create dtreeviz visualization
viz_model = model(
    dt,
    X,
    y,
    target_name='selling_price',
    feature_names=['km_driven']
)

# Display the tree
viz_model.view()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeRegressor
from dtreeviz import model

# Load the cars dataset
df_cars = pd.read_csv('cars.csv')

# Convert owner into a numeric feature
# Example: First Owner -> 0, Second Owner -> 1, etc.
df_cars['owner_encoded'] = df_cars['owner'].astype('category').cat.codes

# Select two input features
X = df_cars[['km_driven', 'owner_encoded']]

# Select target variable
y = df_cars['selling_price']

# Create Decision Tree Regressor
dt = DecisionTreeRegressor(
    max_depth=3,
    criterion='absolute_error',
    random_state=42
)

# Train the model
dt.fit(X, y)

# Create dtreeviz visualization
viz_model = model(
    dt,
    X,
    y,
    target_name='selling_price',
    feature_names=['KM Driven', 'Owner']
)

# Display the tree horizontally
viz_model.view(
    orientation='LR',
    scale=1.5
)
```
