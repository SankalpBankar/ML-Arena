```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
X = np.random.rand(100, 1) - 0.5
y = 3 * X[:,0] ** 2 + 0.05 * np.random.randn(100)
```

```python
df = pd.DataFrame()
```

```python
df['X'] = X.reshape(100)
df['y'] = y
```

```python
df
```

```python
plt.scatter(df['X'], df['y'])
plt.title('X vs Y')
```

```python
df['pred1'] = df['y'].mean()
```

```python
df
```

```python
df['res1'] = df['y'] - df['pred1']
```

```python
df
```

```python
plt.scatter(df['X'], df['y'])
plt.plot(df['X'], df['pred1'], color='red')
```

```python
# Import DecisionTreeRegressor
from sklearn.tree import DecisionTreeRegressor
tree1 = DecisionTreeRegressor(max_leaf_nodes=8)
tree1.fit(df['X'].values.reshape(100,1), df['res1'].values)
```

```python
from sklearn.tree import plot_tree
plot_tree(tree1)
plt.show()
```

```python
# Generating X_test
X_test = np.linspace(-0.5, 0.5, 500)
```

```python
y_pred = 0.265458 + tree1.predict(X_test.reshape(500,1))
```

```python
plt.figure(figsize=(14,4))
plt.subplot(121)
plt.plot(X_test, y_pred, linewidth=2, color='red')
plt.scatter(df['X'], df['y'])
```

```python
df['pred2'] = 0.265458 + tree1.predict(df['X'].values.reshape(100,1))
```

```python
df
```

```python
df['res2'] = df['y'] - df['pred2']
```

```python
df
```

```python
tree2 = DecisionTreeRegressor(max_leaf_nodes=8)
tree2.fit(df['X'].values.reshape(100,1), df['res2'].values)
```

```python
y_pred = 0.265458 + sum(regressor.predict(X_test.reshape(-1, 1)) for regressor in [tree1,tree2])
```

```python
plt.figure(figsize=(14,4))
plt.subplot(121)
plt.plot(X_test, y_pred, linewidth=2, color='red')
plt.scatter(df['X'], df['y'])
plt.title('X vs y')
```

```python
def gradient_boost(X, y, number, lr, count=1, regs=None, foo=None):
    # Initialize list of regressors
    if regs is None:
        regs = []

    # Base case
    if number == 0:
        return regs

    # Compute residuals
    if count > 1:
        residual = y - lr * regs[-1].predict(X)
    else:
        residual = y
        foo = y

    # Train a weak learner
    tree_reg = DecisionTreeRegressor(max_depth=5, random_state=42)
    tree_reg.fit(X, residual)

    # Store the model
    regs.append(tree_reg)

    # Predict using all trees
    x1 = np.linspace(-0.5, 0.5, 500).reshape(-1, 1)

    y_pred = np.zeros(len(x1))
    for reg in regs:
        y_pred += lr * reg.predict(x1)

    # Plot
    print(f"Iteration: {count}")

    plt.figure(figsize=(8, 5))
    plt.plot(x1, y_pred, 'b-', linewidth=2, label='Boosted Prediction')
    plt.scatter(X[:, 0], foo, color='red', s=20, label='Training Data')
    plt.legend()
    plt.grid(True)
    plt.show()

    # Recursive call
    return gradient_boost(
        X,
        y,
        number - 1,
        lr,
        count=count + 1,
        regs=regs,
        foo=foo
    )
```

```python
np.random.seed(42)

X = np.random.rand(100, 1) - 0.5
y = 3 * X[:, 0]**2 + 0.05 * np.random.randn(100)

gradient_boost(X, y, number=5, lr=1)
```
