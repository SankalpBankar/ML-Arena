```python
import random
import numpy as np
```

```python
class KMeans:

    # Initialize the K-Means algorithm
    def __init__(self, n_clusters=2, max_iter=100):

        # Number of clusters we want to create
        self.n_clusters = n_clusters

        # Maximum number of iterations
        self.max_iter = max_iter

        # Centroids will be initialized later
        self.centroids = None


    def fit_predict(self, X):

        # Randomly select 'n_clusters' data points as initial centroids
        random_index = random.sample(
            range(0, X.shape[0]),
            self.n_clusters
        )

        # Store the randomly selected data points as centroids
        self.centroids = X[random_index]


        # Repeat the clustering process for maximum 'max_iter' times
        for i in range(self.max_iter):

            # ------------------------------------------------
            # Step 1: Assign each data point to the nearest
            # centroid
            # ------------------------------------------------
            cluster_group = self.assign_clusters(X)


            # Store the old centroids before moving them
            old_centroids = self.centroids.copy()


            # ------------------------------------------------
            # Step 2: Calculate new centroids
            # ------------------------------------------------
            self.centroids = self.move_centroids(
                X,
                cluster_group
            )


            # ------------------------------------------------
            # Step 3: Check whether the centroids have stopped
            # moving. If they haven't changed, stop the loop.
            # ------------------------------------------------
            if (old_centroids == self.centroids).all():
                break


        # Return the cluster assigned to each data point
        return cluster_group
```

```python
def assign_clusters(self, X):
    cluster_group = []
    distances = []

    for row in X:
        for centroid in self.centroids:
            distances.append(np.sqrt(np.dot(row-centroid, row-centroid)))
```

```python
def move_centroids(self,X,cluster_group):
        new_centroids = []

        cluster_type = np.unique(cluster_group)

        for type in cluster_type:
            new_centroids.append(X[cluster_group == type].mean(axis=0))

        return np.array(new_centroids)
```
