```python
import time
import requests
import pandas as pd

session = requests.Session()
adapter = requests.adapters.HTTPAdapter(max_retries=3)
session.mount("https://", adapter)
session.mount("http://", adapter)

all_pages = []
max_retries = 3

for page in range(1, 429):
    response = None

    for attempt in range(max_retries):
        try:
            response = session.get(
                "https://api.themoviedb.org/3/movie/top_rated",
                params={
                    "api_key": "8265bd1679663a7ea12ac168da84d2e8",
                    "language": "en-US",
                    "page": page,
                },
                timeout=20,
            )
            response.raise_for_status()
            break
        except requests.exceptions.RequestException as err:
            print(f"Attempt {attempt + 1} failed for page {page}: {err}")
            if attempt < max_retries - 1:
                time.sleep(2)
            else:
                print(f"Skipping page {page} after repeated failures.")

    if response is None or response.status_code != 200:
        continue

    page_results = response.json().get("results", [])
    if not page_results:
        break

    temp_df = pd.DataFrame(page_results)[
        ["id", "title", "overview", "release_date", "popularity", "vote_average", "vote_count"]
    ]
    all_pages.append(temp_df)

if all_pages:
    df = pd.concat(all_pages, ignore_index=True)
else:
    df = pd.DataFrame()

print(f"Loaded {len(all_pages)} pages")
df.head()
```
